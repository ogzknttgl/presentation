```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('delta-wire.csv')
x = df['training']
y = df['productivity']

x_mean = np.mean(x)
y_mean = np.mean(y)

num = np.sum((x-x_mean)*(y-y_mean))
den = np.sum((x-x_mean)**2)
beta1 = num/den
beta0 = y_mean - beta1*x_mean

#Scatter plot of X,Y
plt.figure(figsize=(6,4))
plt.scatter(x,y,alpha=0.7)
plt.xlabel('x')
plt.ylabel('y')
plt.title('x vs y')

x_draw = np.linspace(0,4000)
y_draw = beta0 + beta1*x_draw
plt.plot(x_draw,y_draw,color='red')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('baseball_payroll.csv')
x = df['Average Payroll (Million)']
y = df['Total Wins']

x_mean = np.mean(x)
y_mean = np.mean(y)

num = np.sum((x-x_mean)*(y-y_mean))
den = np.sum((x-x_mean)**2)
beta1 = num/den
beta0 = y_mean - beta1*x_mean

#Scatter plot of X,Y
plt.figure(figsize=(6,4))
plt.scatter(x,y,alpha=0.7)
plt.xlabel('x')
plt.ylabel('y')
plt.title('x vs y')

x_draw = np.linspace(25,225)
y_draw = beta0 + beta1*x_draw
plt.plot(x_draw,y_draw,color='red')

y_pred = beta0 + beta1*x
SST = np.sum((y-y_mean)**2)
SSR = np.sum((y_pred-y_mean)**2)

R2 = SSR/SST
print(R2)
```
