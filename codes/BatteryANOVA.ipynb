```python
import numpy as np
import matplotlib.pyplot as plt


def calc_F_stat (lfp, nmc, nca, k, N):
    pooled_data = np.concatenate([lfp,nmc,nca])
    lfpMean = np.mean(lfp)
    nmcMean = np.mean(nmc)
    ncaMean = np.mean(nca)
    grandMean = np.mean(pooled_data)

    SSBetween =NofGData*(lfpMean-grandMean)**2+NofGData*(nmcMean-grandMean)**2+NofGData*(ncaMean-grandMean)**2
    SSTotal = 0
    for j in range(N):
        SSTotal+=(pooled_data[j]-grandMean)**2
    SSWithin = SSTotal-SSBetween
    MSBetween = SSBetween/(k-1)
    MSWithin = SSWithin/(N-k)
    F = MSBetween/MSWithin
    return F
    
k = 3
NofGData = 4
N= k*NofGData
NoofPerm = 10000

lfp, nmc, nca = np.array([28,30,29,31]), np.array([22,24,25,23]), np.array([20,21,23,20])
pooled_data = np.concatenate([lfp,nmc,nca])


F_observed = calc_F_stat (lfp, nmc, nca, k, N)
print(f"F-statistic is {F_observed:.2f}")
FList =[]
for i in range(NoofPerm):
    np.random.shuffle(pooled_data)
    sample_lfp = pooled_data[:NofGData]
    sample_nmc = pooled_data[NofGData:2*NofGData]
    sample_nca = pooled_data[2*NofGData:]
    F = calc_F_stat (sample_lfp, sample_nmc, sample_nca, k, N)
    FList.append(F)

F_Array = np.array(FList)
p_value = np.mean(F_Array>F_observed)
print("Corresponding p-value is ",p_value)  
    




```
